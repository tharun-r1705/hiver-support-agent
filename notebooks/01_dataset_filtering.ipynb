```python
amazon_df = df[df["author_id"] == "AmazonHelp"].copy()

# Create lookup for all tweets
tweet_lookup = df.set_index("tweet_id")

# Get parent tweet information
amazon_df["parent_text"] = amazon_df[
    "in_response_to_tweet_id"
].map(tweet_lookup["text"])

amazon_df["parent_author"] = amazon_df[
    "in_response_to_tweet_id"
].map(tweet_lookup["author_id"])

# Create a thread ID
amazon_df["thread_id"] = amazon_df["in_response_to_tweet_id"]

print(amazon_df[
    [
        "thread_id",
        "parent_author",
        "parent_text",
        "tweet_id",
        "text"
    ]
].head(10).to_string())
```

```python
print("Amazon tweets:", len(amazon_df))
print("Unique threads:", amazon_df["thread_id"].nunique())
```

```python
threads = amazon_df[
    amazon_df["in_response_to_tweet_id"].notna()
].copy()

threads = threads[
    [
        "in_response_to_tweet_id",
        "parent_author",
        "parent_text",
        "tweet_id",
        "author_id",
        "text",
        "response_tweet_id"
    ]
].rename(columns={
    "in_response_to_tweet_id": "customer_tweet_id",
    "parent_author": "customer_author",
    "parent_text": "customer_text",
    "tweet_id": "reply_tweet_id",
    "author_id": "brand",
    "text": "brand_reply"
})

print(threads.shape)
print(threads.head(10).to_string())
```

```python
print("Total interactions:", len(threads))
print("Unique customer tweets:", threads["customer_tweet_id"].nunique())
print("Unique Amazon replies:", threads["reply_tweet_id"].nunique())
```

```python
# Start with AmazonHelp replies that have a parent tweet
threads = amazon_df[
    amazon_df["in_response_to_tweet_id"].notna()
].copy()

# The customer/parent tweet
threads["customer_tweet_id"] = threads["in_response_to_tweet_id"].astype("Int64")

# Create a thread ID
threads["thread_id"] = threads["customer_tweet_id"]

# Keep useful columns
threads = threads[
    [
        "thread_id",
        "customer_tweet_id",
        "parent_author",
        "parent_text",
        "tweet_id",
        "text",
        "response_tweet_id"
    ]
].rename(columns={
    "parent_author": "customer_author",
    "parent_text": "customer_text",
    "tweet_id": "reply_tweet_id",
    "text": "brand_reply"
})

print(threads.shape)
print(threads["thread_id"].nunique())
```

```python
thread_counts = threads["thread_id"].value_counts()

print(thread_counts.head(10))
```

```python
thread_id = thread_counts.index[0]

print(
    threads[threads["thread_id"] == thread_id][
        ["customer_text", "brand_reply", "response_tweet_id"]
    ].to_string(index=False)
)
```

```python
threads_clean = threads.dropna(
    subset=["customer_text", "brand_reply"]
).copy()

print("After removing incomplete records:", threads_clean.shape)
```

```python
threads_clean = threads_clean.drop_duplicates(
    subset=["customer_tweet_id", "reply_tweet_id"]
).copy()

print("After removing duplicates:", threads_clean.shape)
```

```python
sample_size = min(30000, len(threads_clean))

threads_sample = threads_clean.sample(
    n=sample_size,
    random_state=42
).reset_index(drop=True)

print("Final dataset:", threads_sample.shape)
```

```python
from pathlib import Path

Path("D:\\hiver-support-agent\\data\\processed").mkdir(
    parents=True,
    exist_ok=True
)
```

```python
output_path = "D:/hiver-support-agent/data/processed/threads_sample.parquet"

threads_sample.to_parquet(
    output_path,
    index=False,
    engine="fastparquet"
)

print(f"Saved to: {output_path}")
```
